```python
'''Predict student marks based on number_of_hours_study and attendence.'''
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split

df=pd.read_csv('/content/drive/MyDrive/Data set/Student_exam_scores.csv')
df


```

```python
x=df[['hours_studied','attendance_percent','previous_scores']]
y=df['exam_score']
```

```python
mlr=LinearRegression()
mlr.fit(x,y)
```

```python
print("Intercept value is :-",mlr.intercept_)
print("Slope value is :-",mlr.coef_)
```

```python
hours=float(input("Enter the hours student study:-"))
attendance=float(input("Enter the Attendance:-"))
previous_score=float(input("Enter the previous score:-"))
input_data=[[hours,attendance,previous_score]]
predicted_score=mlr.predict(input_data)
print("Predicted exam score is :-",predicted_score[0])
```

```python
plt.figure(figsize=(8,6))
plt.scatter(y,mlr.predict(x),color='blue')
plt.plot([min(y),max(y)],[min(y),max(y)],color='black')
plt.xlabel("Actual marks")
plt.ylabel("Predicted Marks")
plt.title(" Student Marks prediction based on number of hours study and attendance using multiple Linear Regression")
plt.grid()
plt.show()
```
