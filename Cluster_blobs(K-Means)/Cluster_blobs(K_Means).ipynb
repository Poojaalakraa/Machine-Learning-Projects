```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
df=pd.read_csv('/content/drive/MyDrive/Data set/cluster_blobs.csv')
df
```

```python
kmeans=KMeans(n_clusters=3,random_state=42)
df['cluster']=kmeans.fit_predict(df[['X1','X2']])
df
```

```python
print("Cluster centers:-\n")
print(kmeans.cluster_centers_)
```

```python
# data visualization:
plt.figure(figsize=(6,6))
plt.scatter(df[df['cluster']==0]['X1'],df[df['cluster']==0]['X2'],color='red',label='Cluster 1')
plt.scatter(df[df['cluster']==1]['X1'],df[df['cluster']==1]['X2'],color='blue',label='Cluster 2')
plt.scatter(kmeans.cluster_centers_[0][0],kmeans.cluster_centers_[0][1],s=200,c='green',label='Centroid 1')
plt.scatter(kmeans.cluster_centers_[1][0],kmeans.cluster_centers_[1][1],s=200,c='violet',label="Centroid 2")
plt.xlabel('X1')
plt.ylabel('X2')
plt.title('K-means Clustering')
plt.legend()
plt.grid()
plt.show()
```
