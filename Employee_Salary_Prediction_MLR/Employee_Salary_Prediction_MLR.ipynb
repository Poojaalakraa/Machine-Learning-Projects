```python
'''Predict employee salary based on age of employee and years of experience.'''
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
df=pd.read_csv('/content/drive/MyDrive/Data set/Employee salary.csv')
df
```

```python
x=df[['Age','Experience_Years']]
y=df['Monthly_Salary']
```

```python
mlr=LinearRegression()
mlr.fit(x,y)
```

```python
Age=float(input("Enter the Employee age:-"))
Experience_Years=float(input("Enter the Employee Experience:-"))
input_data=[[Age,Experience_Years]]
predicted_score=mlr.predict(input_data)
print("Predicted Salary is :-",predicted_score[0])
```

```python
plt.figure(figsize=(7,6))
plt.scatter(y,mlr.predict(x),color='blue')
plt.plot([min(y),max(y)],[min(y),max(y)],color='black')
plt.xlabel("Actual Salary")
plt.ylabel("Predicted Salary")
plt.title(" Employee Salary prediction based on Years of Experience and age  using multiple Linear Regression")
plt.grid()
plt.show()
```
