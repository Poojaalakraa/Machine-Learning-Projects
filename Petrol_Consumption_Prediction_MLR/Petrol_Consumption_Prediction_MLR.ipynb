```python
''' Predict petrol consumption based on driver license , petrol tax, paved highways and average income.'''
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
df=pd.read_csv('/content/drive/MyDrive/Data set/petrol_consumption.csv')
df
```

```python
x=df[['Petrol_tax','Average_income','Paved_Highways','Population_Driver_licence(%)']]
y=df['Petrol_Consumption']
```

```python

mlr=LinearRegression()
mlr.fit(x,y)
```

```python
print("Intercept value is :-",mlr.intercept_)
print("Slope value is :-",mlr.coef_)
```

```python
Petrol_tax=float(input("Enter the Petrol tax:-"))
average_income=float(input("Enter the average income:-"))
paved_highways=float(input("Enter the paved highways:-"))
population_driver_licence=float(input("Enter the Population_Driver_licence(%):-"))
input_data=[[Petrol_tax,average_income,paved_highways,population_driver_licence]]
new_consumption=[[Petrol_tax,average_income,paved_highways,population_driver_licence]]
predicted_consumption=mlr.predict(new_consumption)
print("Predicted Petrol Consumption is :-",predicted_consumption[0])
```

```python
plt.figure(figsize=(7,6))
plt.scatter(y,mlr.predict(x),color='blue')
plt.plot([min(y),max(y)],[min(y),max(y)],color='black')
plt.xlabel("Actual Total consumption")
plt.ylabel("Predicted Consumption")
plt.title(" Petrol Consumption prediction  using multiple Linear Regression")
plt.grid()
plt.show()
```
