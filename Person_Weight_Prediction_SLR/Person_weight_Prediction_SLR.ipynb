```python
''' Person weight prediction bases on height'''
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
df=pd.read_csv('/content/drive/MyDrive/Data set/height_weight.csv')
df
```

```python
x=df[['Height']]
y=df['Weight']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
user_input=int(input("Enter height in cm:-"))
prediction=lr.predict([[user_input]])
print("Predicted weight ",prediction[0])
```

```python
plt.scatter(x_train,y_train,label='Actual Data',color='Orange')
plt.plot(x_train,lr.predict(x_train),color='yellow',label='Regression Line',linewidth=3)
plt.xlabel('Height(cm)')
plt.ylabel('Weight(kg)')
plt.title("Weight Prediction based on height")
plt.legend()
plt.show()
```
